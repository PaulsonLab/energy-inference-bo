# 1. Question

Does the frozen two-regime system contain a meaningful nonmyopic planning opportunity, and—only if it does—can a shared causal policy and causal path-KL transport recover it?

# 2. Why this model system

The model isolates mechanistic uncertainty in one binary latent regime. Two exploit regions have modest regime separation, while a lower-value diagnostic region is much more informative. The benchmark is a mechanism kill test, not a real-world flagship.

# 3. Frozen model

The terminal utility is best observed improvement. The causal sufficient state is remaining horizon, posterior regime log-odds, and current incumbent. The candidate family and all gates were fixed before any policy optimization.
